```python

print("-"*100)
print("                           WhatsApp Chat Analyer ")
print("                           *********************")
print("-"*100)




print("                             Total Lines")
print("                             -----------")
with open("hostel_bois.txt", "r", encoding="utf-8") as file:
    content = file.read()

lines = content.split("\n")

print("Total Lines:", len(lines))
print("*"*100)


print("                             Total Messages")
print("                             --------------")

messages = []

for line in lines:

    if len(line.strip()) == 0:
        continue

    parts = line.split(" - ", 1)

    if len(parts) < 2:
        continue

    timestamp = parts[0]
    person = parts[1]

    if ": " not in person:
        continue

    sender, message = person.split(": ", 1)

    messages.append({
        "timestamp": timestamp,
        "sender": sender,
        "message": message
    })

print("Total Messages:", len(messages))

print("*"*100)


print("                            Total Participants")
print("                            ------------------")

participants = set()
message_count = {}

for msg in messages:

    sender = msg["sender"]

    participants.add(sender)

    if sender in message_count:
        message_count[sender] += 1
    else:
        message_count[sender] = 1

print("Total Participants:", len(participants))
print()

sorted_count = sorted(message_count.items(), key=lambda x: x[1], reverse=True)

print("Messages Per Person")
print("-" * 25)

for sender, count in sorted_count:
    print(sender, ":", count)

print("*"*100)

print("                               Most Active Day & Hour")
print("                               ----------------------")

day_count = {}
hour_count = {}

for msg in messages:

    date, time = msg["timestamp"].split(", ")
    hour = time[:2]

    day_count[date] = day_count.get(date, 0) + 1
    hour_count[hour] = hour_count.get(hour, 0) + 1

print("Most Active Day :", max(day_count, key=day_count.get))
print("Most Active Hour:", max(hour_count, key=hour_count.get) + ":00")
print("*"*100)

print("                              Top Most used words ")
print("                              --------------------")

word_count = {}

for msg in messages:
    words = msg["message"].split()

    for word in words:
        word = word.lower()
        word_count[word] = word_count.get(word, 0) + 1

top_words = sorted(word_count.items(), key=lambda x: x[1], reverse=True)

print(top_words[:5])
print("*"*100)

print("                        Most Active Person In The Group")
print("                        --------------------------------")

most_active = max(message_count, key=message_count.get)

print("Most Active Person:", most_active)
print("Messages:", message_count[most_active])

print("*"*100)


print("                                 Bar Graph ")
print("                                 ---------")
import matplotlib.pyplot as plt

names = list(message_count.keys())
counts = list(message_count.values())

plt.bar(names, counts)

plt.title("Messages Per Person")
plt.xlabel("Participants")
plt.ylabel("Number of Messages")

plt.show()

print("*"*100)




print("                               Personality Detection")
print("                               ---------------------")
for person, count in message_count.items():

    if count > 500:
        personality = "Spammer"

    elif count > 200:
        personality = "Active"

    else:
        personality = "Silent"

    print(person, ":", personality)
```
